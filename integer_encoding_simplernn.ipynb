```python
import numpy as np
docs = [
    'go india go',
    'have a bright day',
    'the sky is blue',
    'the cat sits right here',
    'sun shines',
    'the dog barks very loudly today',
    'learning deep neural networks easily',
    'python code runs',
    'artificial intelligence is changing tech quickly',
    'natural language processing',
    'text classification needs good training data',
    'machine learning models process text fast',
    'recurrent neural networks',
    'tokenization breaks text',
    'word embeddings capture deep semantic meaning'
]
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer = Tokenizer(oov_token='<nothing>')
```

```python
tokenizer.fit_on_texts(docs)
```

```python
tokenizer.word_index
```

```python
tokenizer.word_counts
```

```python
tokenizer.document_count
```

```python
sequences = tokenizer.texts_to_sequences(docs)
sequences
```

```python
from keras.utils import pad_sequences
sequences = pad_sequences(sequences,padding='post')
sequences
```
